```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("customers.csv")
X = data[['Annual_Income','Spending_Score']].values

# Initial centroids
centroids = np.array([[15,39],[90,90]])

def distance(a,b):
    return np.sqrt(np.sum((a-b)**2))

# Run iterations
for i in range(5):
    clusters = [[],[]]

    for point in X:
        d1 = distance(point, centroids[0])
        d2 = distance(point, centroids[1])

        if d1 < d2:
            clusters[0].append(point)
        else:
            clusters[1].append(point)

    centroids = np.array([np.mean(cluster, axis=0) for cluster in clusters])

# Assign final cluster labels
labels = []
for point in X:
    d1 = distance(point, centroids[0])
    d2 = distance(point, centroids[1])
    labels.append(0 if d1 < d2 else 1)

# Plot graph
plt.figure(figsize=(7,5))
plt.scatter(X[:,0], X[:,1], c=labels)
plt.scatter(centroids[:,0], centroids[:,1], s=200, marker='X')
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Clustering (Manual Mean Method)")
plt.show()
```
